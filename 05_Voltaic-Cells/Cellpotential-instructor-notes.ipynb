# Impact of Concentrations on Voltaic Cell Potentials

# Instructor Notes

##Lesson Information:
- This activity was piloted during a 75-minute recitation session in the second-to-last week of a second-semester general chemistry course, following instruction on electrochemical work and the Nernst equation.

- The session was led by an undergraduate teaching assistant (TA) with support from three undergraduate learning assistants (LAs) who assisted students and addressed technical issues.

- Approximately 60 students attended. The recitation was held in a fixed-seat classroom, which limited student interaction and made it difficult for the TA and LAs to move around and provide assistance.

## Implementation Notes
- Prior to this activity, students had completed an introductory module on Google Colab, Gemini, and basic Python programming, along with two reaction kinetics activities and a titration curve analysis activity using common household acids.

- The present activity is independent of the other activities and may be implemented in any order. We recommend students complete the introductory activity on Google Colab, Gemini, and basic Python programming beforehand to ensure they have the necessary technical foundation.

- Students were randomly assigned to one of three groups, each investigating a different voltaic cell using a corresponding experimental dataset. Group 1 analyzed a **Cr(s)|Cr$^{3+}$(aq)||Ag$^{+}$(aq)|Ag(s)** cell, Group 2 analyzed a **Cr(s)|Cr$^{3+}$(aq)||Co$^{2+}$(aq)|Co(s)** cell, and Group 3 analyzed a **Zn(s)|Zn$^{2+}$(aq)||Ag$^{+}$(aq)|Ag(s)** cell. All datasets consisted of simulated electrochemical measurements with realistic experimental noise added to mimic authentic laboratory data.

## Formative Assessment
- Students used the provided Google Colab notebook and Gemini for data processing and plotting.
Data interpretation and analysis were completed offline using an accompanying guided-inquiry worksheet (PDF).

- For participation credit, students submitted a shareable link to their completed Colab notebook through a Google Form and uploaded their completed worksheet to Canvas.

## Challenges Encountered

This activity proved more challenging than the previous activities, with many students using the full class period to complete it. The increased difficulty may have been due to its implementation near the end of the semester, when students were preparing for final exams, as well as its emphasis on applying electrochemistry concepts and quantitative analysis. In contrast, the earlier activities were primarily qualitative and exploration-based.

# Overview
This module is designed for students to perform data modeling for a voltaic cell under different conditions. Students will use Google Colab, Gemini, and Python programming language to read, interpret, and analyze provided data.

Written by Dr. Linlin Jensen

Some experimental data are taken from Brosmer, J. L. and Peters, D. G., *J. Chem. Educ.* 2012, 89, 6, 763 - 766, https://doi.org/10.1021/ed2004643

> **AI Use Disclaimer:** Microsoft Copilot was used to improve clarity and organization of the activity’s written instructions.

### Learning Objectives
By the end of this recitation, you will be able to:
* Scientific Skills
  - **Data Interpretation:**: Interpret data to determine how changes in  concentrations influence cell potential.
  - **Graphical Representation:**: Graph and interpret how cell potential ($E_{cell}$) varies with concentrations and with logarithms of reaction quotients ($\text{log}⁡Q$) to identify trends.
  - **Mathematical Modeling:** Use the $E_{cell}$ vs. $\text{log}Q$ graph to determine the number of electrons transferred ($n$), the standard cell potential ($E^{o}_{cell}$), and the equilibrium constant ($K$), then evaluate whether these values match theoretical predictions.
* Cyberinfrastructure Skills
  - **Data Handling:** Import and load a CSV file into a Pandas DataFrame.
  - **Python Programming:**
    - Access and open a notebook in Google Colab
    - Identify and use the key components of the Google Colab interface (code cells, text cells, run button).
    - Read and understand a Python script with comments.
    - Run a Python script and interpret its output.
  - **Data Cleaning/Transformation:** Calculate $Q$ and $\text{log}Q$ using mathematical models.
  - **Data Visualization:** Generate scatter plots using Python libraries (e.g., Pandas, Matplotlib, NumPy/SciPy).
  - **Computational Analysis:** Perform a linear fit to find the slope and intercept.
  - **AI Literacy:** Formulate effective and specific prompts to get the most out of an AI assistant.
