# Reaction Kinetics of Organic Reactions

# Instructor Notes

##Lesson Information:
- This activity was piloted during a 75-minute recitation in the second week of a second-semester general chemistry course.

- The session was led by an undergraduate teaching assistant (TA) with support from three undergraduate learning assistants (LAs) who assisted students and addressed technical issues.

- A total of 75 students attended. The recitation was held in a fixed-seat classroom, which limited student interaction and made it difficult for the TA and LAs to move around and provide assistance.

## Implementation Notes
- Students had previously completed an introductory activity on Google Colab, Gemini, and basic Python programming.

- This activity was implemented before integrated rate laws were formally introduced in lecture.

- Students were assigned to one of two groups, each working with a different molecule and experimental dataset. Group 1 analyzed the degradation of ibuprofen, whose simulated data followed first-order kinetics, whereas Group 2 analyzed an aldehyde, whose simulated data followed second-order kinetics.

## Formative Assessment
- Students used the provided Google Colab notebook and Gemini for data processing and plotting.
Data interpretation and analysis were completed offline using an accompanying guided-inquiry worksheet (PDF).

- For participation credit, students submitted a shareable link to their completed Colab notebook through a Google Form and uploaded their completed worksheet to Canvas.

## Challenges Encountered
- Several students did not bring laptops and completed the activity using tablets or smartphones, which limited their ability to work efficiently in Google Colab.

- For some prompts, Gemini did not directly generate the requested code. Instead, it presented students with a series of guided steps or asked them to choose how they would like to proceed, providing only a summary of the completed task rather than the underlying code.

- In other cases, Gemini inserted outputs or results into different sections of the notebook, making it difficult for students to locate the requested code and analyses.

## Potential Workarounds for Inconsistent Gemini Responsens

- Use more explicit prompts, such as: "Provide only the Python code needed to complete this task. Do not provide step-by-step explanations or summaries."

- Ask students to regenerate the response if Gemini provides instructions instead of code.

- Encourage students to start a new chat session when responses become inconsistent.

- Remind students to review the entire notebook for generated outputs.

- Remind students to verify that all generated code appears in the intended notebook cell before executing it.

# Step 2: Calculating and Plotting Concentrations for 1st & 2nd Order Kinetics

Our goal is to calculate the theoretical reactant concentrations at each time point using the 1st and 2nd order integrated rate laws and compare them to the measured values.

## Determine the 1st and 2nd order rate constants

We start by prompting Gemini to calculate the 1st and 2nd order rate constants ($k_1$ and $k_2$), using the estimated half life from the previous step.

The rate constant relate to the half life through the following equations:

<p style='text-align: right;'>
$t_\frac{1}{2} = \dfrac{\text{ln}(2)}{k_1}$  
$\qquad\qquad\qquad\qquad\qquad$ (1st-order Half-life Equation)  
<br>

<p style='text-align: right;'>
$t_\frac{1}{2} = \dfrac{1}{k_2 [A]_0}$  
$\qquad\qquad\qquad\qquad\qquad$   (2nd-order Half-life Equation)   
<br>

In the above equations, $[A]_0$ stands for the initial concentration of the reactant.

# Step 3: Using Linear Forms of Rate Laws to Find Reaction Order

The linear form of the integrated rate laws for 1st and 2nd order reactions are:

<p style='text-align: right;'>
$\text{ln}[A] = -kt + \text{ln} [A]_0$
$\qquad\qquad\qquad\qquad\qquad$  (1st-order Integrated Rate Law)
</p>

<p style='text-align: right;'>
$\frac{1}{[A]} = kt + \frac{1}{[A]_0}$
$\qquad\qquad\qquad\qquad$ (2nd-order Integrated Rate Law)
</p>

In the above equations, $[A]_0$ stands for the initial concentration of the reactant.
